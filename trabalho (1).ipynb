```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv('Manufacturers_Suppliers_Tratado.csv', encoding='latin1', sep=';')
df.head()
```

```python
conn = sqlite3.connect('data/dados_finais.db')
df_man = df[['manufacturer_id', 'manufacturer_name', 'created_date']]
df_man.columns = ['manufacturer_id', 'manufacturer_name', 'manufacturer_date']
df_man.to_sql('Manufacturer', conn, if_exists='replace', index=False)
conn.close()
```

```python
conn = sqlite3.connect('data/dados_finais.db')
df_sup = df[['supplier_id', 'supplier_title', 'supplier_category']].drop_duplicates()
df_sup.columns = ['suplier_id', 'suplier_title', 'suplier_category']
df_sup.to_sql('Suplier', conn, if_exists='replace', index=False)
conn.close()
```

```python
conn = sqlite3.connect('data/dados_finais.db')
df_trans = df[['transaction_date', 'amount', 'manufacturer_id', 'supplier_id']]
df_trans.to_sql('Transaction', conn, if_exists='replace', index=False)
conn.close()
```

```python
import sqlite3

conn = sqlite3.connect('data/dados_finais.db')
df_mod = df[['model_id', 'model_info',]].drop_duplicates()
df_mod.to_sql('Model', conn, if_exists='replace', index=False)
conn.close()
```
